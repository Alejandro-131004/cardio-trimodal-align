```python
# Import Necessary packages
import librosa
import matplotlib.pyplot as plt
import numpy as np
import soundfile as sf
```

```python
# Open PCG file version 1
audio_file = r"C:\Users\Bruno Oliveira\Desktop\DataBase_multiscope_september_2023_plus\1_AV.mp3" # Path to mp3 file
y, sr = librosa.load(audio_file, sr=None) # It resamples the data to 22050 Hz, so its better to use with sr=None

# Create time axis
time = np.linspace(0, len(y)/sr, num=len(y))

# Plot the signal
plt.figure(figsize=(14, 5))
plt.plot(time, y)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Open PCG file version 2
y, sr = sf.read(audio_file) # best option, it preserves original sample rate

# Create time axis
time = np.linspace(0, len(y)/sr, num=len(y))

# Plot the signal
plt.figure(figsize=(14, 5))
plt.plot(time, y)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Open ECG file
with open(r"C:\Users\Bruno Oliveira\Desktop\DataBase_multiscope_september_2023_plus\1_AV.raw") as file:
    data = file.read().split(',')  # Split the contents by comma
    data = [float(i) for i in data]  # Convert strings to floats
    

# Define sample frequency
fs = 500 # Hz

# Create time axis
t = np.arange(len(data)) / fs

plt.figure(figsize=(40,10))
plt.plot(t, data)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('ECG signal')
plt.show()
```
